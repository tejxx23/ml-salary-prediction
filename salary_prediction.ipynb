# 💰 Salary Prediction using Multiple Linear Regression

## Project Overview

This project uses Machine Learning to predict an employee's salary
based on their:

- Experience
- Age
- Hours Worked

A Multiple Linear Regression model is trained and evaluated using
MAE, MSE, RMSE, and R² Score.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

## 2. Create Dataset

```python
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5, 6, 7, 8],
    "Age": [22, 24, 26, 29, 31, 34, 36, 40],
    "Hours_Worked": [40, 42, 44, 45, 46, 48, 50, 52],
    "Salary": [30000, 34000, 39000, 45000, 50000, 56000, 62000, 70000]
})

df
```

## 3. Select Features and Target

```python
X = df[["Experience", "Age", "Hours_Worked"]]
Y = df["Salary"]

print("Features:")
print(X)

print("\nTarget:")
print(Y)
```

## 4. Train-Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

print("Training data:", len(X_train))
print("Testing data:", len(X_test))
```

## 5. Train Linear Regression Model

```python
model = LinearRegression()

model.fit(X_train, Y_train)
```

## 6. Make Predictions

```python
predictions = model.predict(X_test)

print("Actual values:", Y_test.values)
print("Predicted values:", predictions)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

## 7. Model Evaluation

```python
mae = mean_absolute_error(Y_test, predictions)
mse = mean_squared_error(Y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, predictions)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

## 8. Predict Salary for a New Employee

```python
new_employee = pd.DataFrame({
    "Experience": [6],
    "Age": [30],
    "Hours_Worked": [48]
})

predicted_salary = model.predict(new_employee)

print("Predicted Salary: ₹", round(predicted_salary[0], 2))
```

## 9. Conclusion

A Multiple Linear Regression model was used to predict employee salary
based on Experience, Age, and Hours Worked.

The model was evaluated using MAE, MSE, RMSE, and R² Score.

The trained model can also predict the salary of a new employee
based on their input features.
